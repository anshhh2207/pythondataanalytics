```python
import pandas as pd
```

```python
def Category(Age):
    if Age<25:
        return'young'
    elif Age>=25& Age <=35:
        return'Adult'
    else:
        return 'Senior'
df['Age_Category'] = df['Age'].apply(Category)
df
```

```python
def status(BusinessTravel):
    if BusinessTravel=='Travel_Rarely':
        return'Regular'
    elif BusinessTravel=='Travel_Frequently':
        return'Frequent'
    else:
        return 'No travel'
df['Travel_Status'] = df['BusinessTravel'].apply(status)
df
```

```python
def Astatus(Attrition):
    if Attrition=='Yes':
        return'Left Company'
    else:
        return 'Working'
df['Travel_Status'] = df['Attrition'].apply(Astatus)
df
```

```python
df['Department'] = df['Department'].apply(lambda x: x.upper())
df
```

```python
df["DailyRate_Bonus"] = df["DailyRate"] * 1.10
df
```

```python
def dcategory(DistanceFromHome):
    if DistanceFromHome<=5:
        return 'Near'
    elif DistanceFromHome <=15:
        return 'Medium'
    else:
        return 'Far'
df['Distance_Category'] = df['DistanceFromHome'].apply(dcategory)
df
```

```python
def experience_status(YearsAtCompany):
    if YearsAtCompany >= 10:
        return "Experienced"
    else:
        return "Fresher"
df["Experience_Status"] = df["YearsAtCompany"].apply(experience_status)
df
```
