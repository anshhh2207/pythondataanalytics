```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_excel(r"C:\Users\ANSHIKA\Downloads\d4bdb23b-4858-482e-9d51-d9fa09faeec4.xlsx")
df
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()

plt.show()
```

```python
plt.bar(df["Month"], df["Profit"])

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.show()
```

```python
plt.scatter(df["Marketing_Spend"], df["Sales"])

plt.title("Marketing Spend vs Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")
plt.grid()

plt.show()
```

```python
plt.hist(df["Sales"], bins=6)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 8))
# Monthly Sales
plt.subplot(2, 2, 1)
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
# Monthly Profit
plt.subplot(2, 2, 2)
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
# Marketing Spend vs Sales
plt.subplot(2, 2, 3)
plt.scatter(df["Marketing_Spend"], df["Sales"])
plt.title("Marketing Spend vs Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")
plt.grid()
# Monthly Customers
plt.subplot(2, 2, 4)
plt.plot(df["Month"], df["Customers"], marker="o")
plt.title("Monthly Customers")
plt.xlabel("Month")
plt.ylabel("Customers")
plt.grid()
plt.tight_layout()
plt.show()
```
