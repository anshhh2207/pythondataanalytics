```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\ANSHIKA\OneDrive\Documents\details.csv",header=None)
df
```

```python
df.columns = ["Emp_ID", "Name", "Age", "Gender", "Salary", "Department", "Joining_Date"]
df
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df
```

```python
df["Name"] = df["Name"].fillna("Unknown")
df
```

```python
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
df
```

```python
df["Joining_Date"] = df["Joining_Date"].fillna("2023-01-01")
df
```

```python
df.isnull().sum()
```

```python
df["Emp_ID"] = df["Emp_ID"].astype(int)
df.dtypes
```

```python
df["Age"] = df["Age"].astype(int)
df.dtypes
```

```python
df["Salary"] = df["Salary"].astype(float)
df.dtypes
```

```python
df["Gender"] = df["Gender"].astype("category")
df.dtypes
```

```python
df["Department"] = df["Department"].astype("category")
df.dtypes
```
